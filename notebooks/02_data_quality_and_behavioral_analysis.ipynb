```python
# Summarizing filtered product coverage and key missing product fields by category

filtered_category_products_summary_query = f"""
SELECT
    category_name,
    COUNT(*) AS filtered_product_count,
    COUNTIF(TRIM(COALESCE(product_title, "")) = "") AS blank_product_title_count,
    COUNTIF(TRIM(COALESCE(main_category, "")) = "") AS blank_main_category_count,
    COUNTIF(TRIM(COALESCE(store, "")) = "") AS blank_store_count,
    COUNTIF(price_amount IS NULL) AS missing_price_count
FROM `{configuration["gcp_project_id"]}.{configuration["bigquery_core_dataset_id"]}.filtered_category_products`
GROUP BY category_name
ORDER BY category_name
"""

filtered_category_products_summary_dataframe = run_query(filtered_category_products_summary_query)
display(filtered_category_products_summary_dataframe)
```

```python
# Comparing original and filtered product counts to measure retained product coverage

filtered_vs_original_products_query = f"""
SELECT
    original.category_name,
    original.original_product_count,
    filtered.filtered_product_count,
    original.original_product_count - filtered.filtered_product_count AS excluded_product_count,
    SAFE_DIVIDE(
        filtered.filtered_product_count,
        original.original_product_count
    ) AS retained_product_ratio
FROM (
    SELECT
        category_name,
        COUNT(*) AS original_product_count
    FROM `{configuration["gcp_project_id"]}.{configuration["bigquery_core_dataset_id"]}.products`
    GROUP BY category_name
) AS original
LEFT JOIN (
    SELECT
        category_name,
        COUNT(*) AS filtered_product_count
    FROM `{configuration["gcp_project_id"]}.{configuration["bigquery_core_dataset_id"]}.filtered_category_products`
    GROUP BY category_name
) AS filtered
    ON original.category_name = filtered.category_name
ORDER BY original.category_name
"""

filtered_vs_original_products_dataframe = run_query(filtered_vs_original_products_query)
display(filtered_vs_original_products_dataframe)
```

```python
# Checking how well reviews join to the filtered product universe

review_to_filtered_product_join_query = f"""
SELECT
    reviews.category_name,
    COUNT(*) AS review_count,
    COUNTIF(products.parent_asin IS NOT NULL) AS matched_review_count,
    COUNTIF(products.parent_asin IS NULL) AS unmatched_review_count,
    SAFE_DIVIDE(COUNTIF(products.parent_asin IS NOT NULL), COUNT(*)) AS matched_review_ratio
FROM `{configuration["gcp_project_id"]}.{configuration["bigquery_core_dataset_id"]}.reviews` AS reviews
LEFT JOIN `{configuration["gcp_project_id"]}.{configuration["bigquery_core_dataset_id"]}.filtered_category_products` AS products
    ON reviews.category_name = products.category_name
   AND reviews.parent_asin = products.parent_asin
GROUP BY reviews.category_name
ORDER BY reviews.category_name
"""

review_to_filtered_product_join_dataframe = run_query(review_to_filtered_product_join_query)
display(review_to_filtered_product_join_dataframe)
```

```python
# Verifying filtered review coverage and review-to-product join quality by category

filtered_reviews_verification_query = f"""
WITH original_review_counts AS (
    SELECT
        category_name,
        COUNT(*) AS original_review_count,
        COUNT(DISTINCT parent_asin) AS original_distinct_parent_asin_count
    FROM `{configuration["gcp_project_id"]}.{configuration["bigquery_core_dataset_id"]}.reviews`
    GROUP BY category_name
),
filtered_review_counts AS (
    SELECT
        category_name,
        COUNT(*) AS filtered_review_count,
        COUNT(DISTINCT parent_asin) AS filtered_distinct_parent_asin_count
    FROM `{configuration["gcp_project_id"]}.{configuration["bigquery_core_dataset_id"]}.filtered_category_reviews`
    GROUP BY category_name
),
filtered_review_join_check AS (
    SELECT
        reviews.category_name,
        COUNT(*) AS filtered_review_join_count,
        COUNTIF(products.parent_asin IS NULL) AS unmatched_filtered_review_count
    FROM `{configuration["gcp_project_id"]}.{configuration["bigquery_core_dataset_id"]}.filtered_category_reviews` AS reviews
    LEFT JOIN `{configuration["gcp_project_id"]}.{configuration["bigquery_core_dataset_id"]}.filtered_category_products` AS products
        ON reviews.category_name = products.category_name
       AND reviews.parent_asin = products.parent_asin
    GROUP BY reviews.category_name
)

SELECT
    original.category_name,
    original.original_review_count,
    filtered.filtered_review_count,
    original.original_review_count - filtered.filtered_review_count AS excluded_review_count,
    SAFE_DIVIDE(filtered.filtered_review_count, original.original_review_count) AS retained_review_ratio,
    original.original_distinct_parent_asin_count,
    filtered.filtered_distinct_parent_asin_count,
    join_check.unmatched_filtered_review_count
FROM original_review_counts AS original
INNER JOIN filtered_review_counts AS filtered
    ON original.category_name = filtered.category_name
INNER JOIN filtered_review_join_check AS join_check
    ON original.category_name = join_check.category_name
ORDER BY original.category_name
"""


filtered_reviews_verification_dataframe = run_query(filtered_reviews_verification_query)
display(filtered_reviews_verification_dataframe)
```

```python
# Ranking products by dissatisfaction severity and business priority
product_issue_priority_query = f"""
WITH ranked_products AS (
    SELECT
        *,
        ROW_NUMBER() OVER (
            PARTITION BY category_name, issue_priority_level
            ORDER BY review_count DESC, low_rating_ratio DESC, average_rating ASC
        ) AS category_priority_rank
    FROM `{configuration["gcp_project_id"]}.{configuration["bigquery_core_dataset_id"]}.product_issue_priority`
    WHERE issue_priority_level IN ("Highest priority", "Priority review")
)
SELECT *
FROM ranked_products
WHERE category_priority_rank <= 20
ORDER BY category_name, issue_priority_level, category_priority_rank
"""

product_content_gap_actionable_query = f"""
WITH ranked_products AS (
    SELECT
        *,
        ROW_NUMBER() OVER (
            PARTITION BY category_name, metadata_improvement_priority
            ORDER BY review_count DESC, low_rating_ratio DESC, average_rating ASC
        ) AS category_priority_rank
    FROM `{configuration["gcp_project_id"]}.{configuration["bigquery_core_dataset_id"]}.product_content_gap_priority`
    WHERE metadata_improvement_priority IN ("Highest priority", "Priority review")
)
SELECT *
FROM ranked_products
WHERE category_priority_rank <= 20
ORDER BY category_name, metadata_improvement_priority, category_priority_rank
"""

product_content_gap_caution_query = f"""
WITH ranked_products AS (
    SELECT
        *,
        ROW_NUMBER() OVER (
            PARTITION BY category_name, metadata_improvement_priority
            ORDER BY review_count DESC, low_rating_ratio DESC, average_rating ASC
        ) AS category_priority_rank
    FROM `{configuration["gcp_project_id"]}.{configuration["bigquery_core_dataset_id"]}.product_content_gap_priority`
    WHERE metadata_improvement_priority = "Use with caution"
)
SELECT *
FROM ranked_products
WHERE category_priority_rank <= 20
ORDER BY category_name, category_priority_rank
"""


product_issue_priority_dataframe = run_query(product_issue_priority_query)
product_content_gap_actionable_dataframe = run_query(product_content_gap_actionable_query)
product_content_gap_caution_dataframe = run_query(product_content_gap_caution_query)

display(product_issue_priority_dataframe)
display(product_content_gap_actionable_dataframe)
display(product_content_gap_caution_dataframe)
```
