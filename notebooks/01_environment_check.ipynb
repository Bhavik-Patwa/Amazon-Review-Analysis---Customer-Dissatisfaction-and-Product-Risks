```python
from pathlib import Path
import os

from dotenv import load_dotenv
from google.cloud import bigquery
```

```python
project_root = Path.cwd().resolve().parent if Path.cwd().name == "notebooks" else Path.cwd().resolve()
env_file_path = project_root / ".env"

load_dotenv(dotenv_path = env_file_path)

print(f"Project root : {project_root}")
print(f".env file found : {env_file_path.exists()}")
```

```python
gcp_project_id = os.getenv("GCP_PROJECT_ID")
credentials_path_value = os.getenv("GOOGLE_APPLICATION_CREDENTIALS")
bigquery_location = os.getenv("BIGQUERY_LOCATION")

print(f"GCP project ID : {gcp_project_id}")
print(f"Credentials path from .env : {credentials_path_value}")
print(f"BigQuery location : {bigquery_location}")
```

```python
credentials_file_path = (project_root / credentials_path_value).resolve()

print(f"Resolved credentials file path : {credentials_file_path}")
print(f"Credentials file exists : {credentials_file_path.exists()}")
```

```python
os.environ["GOOGLE_APPLICATION_CREDENTIALS"] = str(credentials_file_path)
print("GOOGLE_APPLICATION_CREDENTIALS has been set for this notebook session.")
```

```python
client = bigquery.Client(project = gcp_project_id)
print("BigQuery client created successfully.")
print("BigQuery project :", client.project)
print("BigQuery location :", bigquery_location)
```

```python
test_query = """
    SELECT
        'BigQuery connection successful' AS status,
        CURRENT_DATE() AS current_date
"""

result_dataframe = client.query(test_query).to_dataframe()
result_dataframe
```
